# Rich Print Examples

Демонстрация использования `rich.print` для красивого вывода данных.


```python
# Импорт rich
from rich import print as rprint
from rich.console import Console
from rich.text import Text
from rich.panel import Panel
from rich.columns import Columns
from rich.table import Table
from rich.progress import Progress, SpinnerColumn, TextColumn, BarColumn, TimeElapsedColumn
from rich.syntax import Syntax
from rich.markdown import Markdown
from rich.tree import Tree
import time
```

## 1. Базовое использование rich.print


```python
# Базовые примеры rich.print
rprint("Это [bold red]красивый[/bold red] текст с [blue]цветами[/blue]!")

# Различные стили
rprint("[bold]Жирный текст[/bold]")
rprint("[italic]Курсив[/italic]")
rprint("[underline]Подчеркнутый[/underline]")
rprint("[strike]Зачеркнутый[/strike]")

# Цвета
rprint("[red]Красный текст[/red]")
rprint("[green]Зеленый текст[/green]")
rprint("[blue]Синий текст[/blue]")
rprint("[yellow]Желтый текст[/yellow]")
rprint("[magenta]Пурпурный текст[/magenta]")
rprint("[cyan]Голубой текст[/cyan]")

# Комбинирование стилей
rprint("[bold red]Жирный красный текст[/bold red]")
rprint("[italic blue]Курсивный синий текст[/italic blue]")
```

## 2. Панели и группировка


```python
# Панели для группировки информации
rprint(Panel("Это простая панель", title="Информация"))

# Панель с различными стилями
rprint(Panel(
    "[bold green]Успешно![/bold green]\nМодель обучена за 2.5 часа\nТочность: 94.2%",
    title="[bold green]Результат обучения[/bold green]",
    border_style="green"
))

# Панель с ошибкой
rprint(Panel(
    "[bold red]Ошибка![/bold red]\nНедостаточно памяти GPU\nТребуется минимум 8GB",
    title="[bold red]Ошибка выполнения[/bold red]",
    border_style="red"
))

# Несколько панелей в колонках
panel1 = Panel("Модель: BERT-base\nПараметры: 110M\nТочность: 92%", title="BERT")
panel2 = Panel("Модель: RoBERTa\nПараметры: 125M\nТочность: 93%", title="RoBERTa")
panel3 = Panel("Модель: GPT-2\nПараметры: 117M\nТочность: 89%", title="GPT-2")

rprint(Columns([panel1, panel2, panel3], equal=True, expand=True))
```

## 3. Синтаксис и код


```python
# Подсветка синтаксиса Python
python_code = '''
def train_model(model, data, epochs=10):
    """Обучение модели машинного обучения"""
    for epoch in range(epochs):
        loss = model.train_step(data)
        print(f"Epoch {epoch}: Loss = {loss:.4f}")
    return model
'''

rprint(Syntax(python_code, "python", theme="monokai", line_numbers=True))

# JSON код
json_code = '''
{
    "model": "BERT-base",
    "parameters": 110000000,
    "accuracy": 0.92,
    "training_time": "2.5 hours"
}
'''

rprint(Syntax(json_code, "json", theme="github"))

# Markdown
markdown_text = """
# Результаты эксперимента

## Модели
- **BERT**: 92% точность
- **RoBERTa**: 93% точность  
- **GPT-2**: 89% точность

## Выводы
Лучшая модель: **RoBERTa** с точностью 93%
"""

rprint(Markdown(markdown_text))
```

## 4. Деревья и иерархия


```python
# Создание дерева структуры проекта
tree = Tree("🤖 ML Project")
tree.add("📁 data/")
tree.add("📁 models/")
tree.add("📁 notebooks/")
tree.add("📁 src/")

# Добавление подэлементов
data_branch = tree.add("📁 data/")
data_branch.add("📄 train.csv")
data_branch.add("📄 test.csv")
data_branch.add("📄 validation.csv")

models_branch = tree.add("📁 models/")
models_branch.add("📄 bert_model.pkl")
models_branch.add("📄 roberta_model.pkl")
models_branch.add("📄 gpt2_model.pkl")

notebooks_branch = tree.add("📁 notebooks/")
notebooks_branch.add("📄 01_data_exploration.ipynb")
notebooks_branch.add("📄 02_model_training.ipynb")
notebooks_branch.add("📄 03_evaluation.ipynb")

src_branch = tree.add("📁 src/")
src_branch.add("📄 train.py")
src_branch.add("📄 evaluate.py")
src_branch.add("📄 utils.py")

rprint(tree)

# Дерево результатов экспериментов
results_tree = Tree("🧪 Результаты экспериментов")
exp1 = results_tree.add("🔬 Эксперимент 1: Baseline")
exp1.add("📊 Точность: 85%")
exp1.add("⏱️ Время: 30 мин")
exp1.add("💾 Память: 2.1 GB")

exp2 = results_tree.add("🔬 Эксперимент 2: + Data Augmentation")
exp2.add("📊 Точность: 89%")
exp2.add("⏱️ Время: 45 мин")
exp2.add("💾 Память: 2.3 GB")

exp3 = results_tree.add("🔬 Эксперимент 3: + Pretraining")
exp3.add("📊 Точность: 94%")
exp3.add("⏱️ Время: 90 мин")
exp3.add("💾 Память: 3.1 GB")

rprint(results_tree)
```

## 5. Прогресс-бары и анимация


```python
# Прогресс-бар для обучения модели
with Progress(
    SpinnerColumn(),
    TextColumn("[progress.description]{task.description}"),
    BarColumn(),
    TextColumn("[progress.percentage]{task.percentage:>3.0f}%"),
    TimeElapsedColumn(),
    console=Console(),
    transient=False
) as progress:
    
    # Симуляция обучения модели
    task = progress.add_task("Обучение модели BERT...", total=100)
    
    for epoch in range(10):
        for step in range(10):
            time.sleep(0.1)  # Симуляция работы
            progress.update(task, advance=1, 
                          description=f"Эпоха {epoch+1}/10, Шаг {step+1}/10")

# Простой прогресс-бар
rprint("📊 Прогресс обучения:")
rprint("Эпоха 1/10: [green]██████████[/green] 100%")
rprint("Эпоха 2/10: [green]██████████[/green] 100%")
rprint("Эпоха 3/10: [green]██████████[/green] 100%")
rprint("Эпоха 4/10: [yellow]████████░░[/yellow] 80%")
rprint("Эпоха 5/10: [red]████░░░░░░[/red] 40%")
```

## 7. Практические примеры для ML


```python
# Функция для красивого вывода результатов обучения
def print_training_results(epoch, loss, accuracy, time_elapsed):
    """Красивый вывод результатов обучения"""
    rprint(f"\n[bold blue]Эпоха {epoch}[/bold blue]")
    rprint(f"📉 [red]Loss: {loss:.4f}[/red]")
    rprint(f"📈 [green]Accuracy: {accuracy:.2%}[/green]")
    rprint(f"⏱️ [yellow]Время: {time_elapsed:.1f}с[/yellow]")
    
    # Условное форматирование
    if accuracy > 0.9:
        rprint("🎉 [bold green]Отличный результат![/bold green]")
    elif accuracy > 0.8:
        rprint("👍 [yellow]Хороший результат[/yellow]")
    else:
        rprint("⚠️ [red]Требует улучшения[/red]")

# Симуляция обучения
epochs = [1, 2, 3, 4, 5]
losses = [2.45, 1.89, 1.34, 0.98, 0.67]
accuracies = [0.65, 0.78, 0.85, 0.91, 0.94]
times = [45.2, 42.1, 38.7, 35.3, 32.8]

for epoch, loss, acc, time_elapsed in zip(epochs, losses, accuracies, times):
    print_training_results(epoch, loss, acc, time_elapsed)

# Финальный отчет
rprint("\n" + "="*50)
rprint("[bold green]🏆 ОБУЧЕНИЕ ЗАВЕРШЕНО![/bold green]")
rprint("="*50)
rprint(f"📊 [bold]Финальная точность:[/bold] [green]{accuracies[-1]:.2%}[/green]")
rprint(f"📉 [bold]Финальные потери:[/bold] [red]{losses[-1]:.4f}[/red]")
rprint(f"⏱️ [bold]Общее время:[/bold] [blue]{sum(times):.1f}с[/blue]")
rprint("="*50)
```

## Заключение

`rich.print` предоставляет мощные возможности для красивого вывода:

### Основные возможности:
- **Стилизация текста**: цвета, жирный, курсив, подчеркивание
- **Панели**: группировка информации в красивые блоки
- **Синтаксис**: подсветка кода Python, JSON, Markdown
- **Деревья**: иерархическое отображение структуры
- **Прогресс-бары**: визуализация выполнения задач
- **Эмодзи**: использование символов для лучшего восприятия

### Ключевые теги стилизации:
- `[bold]` - жирный текст
- `[italic]` - курсив
- `[red]`, `[green]`, `[blue]` - цвета
- `[underline]` - подчеркивание
- `[strike]` - зачеркивание

### Полезные классы:
- `Panel` - для создания панелей
- `Tree` - для иерархических структур
- `Syntax` - для подсветки кода
- `Progress` - для прогресс-баров
- `Markdown` - для markdown разметки
