# Rich Tables Demo

Демонстрация использования библиотеки `rich` для создания красивых таблиц в Python.


```python
# Установка и импорт rich
# %pip install rich

from rich.console import Console
from rich.table import Table
from rich import print as rprint
from rich.panel import Panel
from rich.columns import Columns
from rich.text import Text
from rich.align import Align
import pandas as pd
import numpy as np
```

## 1. Базовое создание таблицы


```python
# Создание консоли для вывода
console = Console()

# Создание простой таблицы
table = Table(title="Модели машинного обучения")

# Добавление колонок
table.add_column("Модель", style="cyan", no_wrap=True)
table.add_column("Точность", style="magenta")
table.add_column("Время обучения", justify="right", style="green")
table.add_column("Размер", justify="right", style="blue")

# Добавление строк
table.add_row("BERT", "0.92", "2.5 часа", "110M")
table.add_row("GPT-2", "0.89", "1.8 часа", "117M")
table.add_row("T5", "0.94", "3.2 часа", "220M")
table.add_row("RoBERTa", "0.93", "2.8 часа", "125M")

# Вывод таблицы
console.print(table)
```

## 2. Стилизация таблиц


```python
# Таблица с различными стилями
table = Table(
    title="[bold blue]Результаты экспериментов[/bold blue]",
    title_style="bold magenta",
    show_header=True,
    header_style="bold red",
    border_style="bright_blue",
    show_lines=True
)

table.add_column("Эксперимент", style="cyan", no_wrap=True, width=20)
table.add_column("F1-Score", style="green", justify="center")
table.add_column("Precision", style="yellow", justify="center")
table.add_column("Recall", style="red", justify="center")
table.add_column("Статус", style="bold")

# Добавление данных с различными стилями
table.add_row("Baseline", "0.85", "0.87", "0.83", "[red]❌ Плохо[/red]")
table.add_row("+ Data Aug", "0.89", "0.91", "0.87", "[yellow]⚠️ Средне[/yellow]")
table.add_row("+ Pretraining", "0.94", "0.95", "0.93", "[green]✅ Отлично[/green]")
table.add_row("+ Fine-tuning", "0.96", "0.97", "0.95", "[bold green]🏆 Превосходно[/bold green]")

console.print(table)
```

## 3. Конвертация из Pandas DataFrame


```python
# Создание DataFrame с данными
data = {
    'Модель': ['BERT-base', 'BERT-large', 'RoBERTa-base', 'RoBERTa-large', 'GPT-2'],
    'Параметры (M)': [110, 340, 125, 355, 117],
    'Точность': [0.92, 0.94, 0.93, 0.95, 0.89],
    'Время (мин)': [45, 120, 50, 140, 35],
    'Память (GB)': [2.1, 6.8, 2.3, 7.2, 1.8]
}

df = pd.DataFrame(data)

# Функция для конвертации DataFrame в Rich Table
def df_to_rich_table(df, title="DataFrame"):
    table = Table(title=title, show_header=True, header_style="bold magenta")
    
    # Добавление колонок
    for col in df.columns:
        table.add_column(col, style="cyan", no_wrap=True)
    
    # Добавление строк
    for _, row in df.iterrows():
        table.add_row(*[str(val) for val in row.values])
    
    return table

# Вывод таблицы
console.print(df_to_rich_table(df, "Сравнение моделей"))
```

## 4. Продвинутые возможности


```python
# Таблица с прогресс-барами и условным форматированием
def create_advanced_table():
    table = Table(
        title="[bold]Мониторинг обучения модели[/bold]",
        show_header=True,
        header_style="bold blue",
        border_style="green"
    )
    
    table.add_column("Эпоха", style="cyan", width=8)
    table.add_column("Loss", style="red", justify="right", width=10)
    table.add_column("Accuracy", style="green", justify="right", width=12)
    table.add_column("Время", style="yellow", width=10)
    table.add_column("Статус", style="bold", width=15)
    
    # Симуляция данных обучения
    epochs = [1, 2, 3, 4, 5]
    losses = [2.45, 1.89, 1.34, 0.98, 0.67]
    accuracies = [0.65, 0.78, 0.85, 0.91, 0.94]
    times = ["2:15", "2:12", "2:08", "2:05", "2:03"]
    
    for epoch, loss, acc, time in zip(epochs, losses, accuracies, times):
        # Условное форматирование на основе значений
        if acc >= 0.9:
            status = "[bold green]✅ Отлично[/bold green]"
        elif acc >= 0.8:
            status = "[yellow]⚠️ Хорошо[/yellow]"
        else:
            status = "[red]❌ Требует улучшения[/red]"
            
        table.add_row(
            str(epoch),
            f"{loss:.2f}",
            f"{acc:.2%}",
            time,
            status
        )
    
    return table

console.print(create_advanced_table())
```

## 5. Группировка и панели


```python
# Создание нескольких таблиц в панелях
def create_model_comparison():
    # Таблица для метрик производительности
    perf_table = Table(title="Производительность", show_header=True, header_style="bold green")
    perf_table.add_column("Модель", style="cyan")
    perf_table.add_column("F1-Score", style="green", justify="right")
    perf_table.add_column("Precision", style="yellow", justify="right")
    perf_table.add_column("Recall", style="red", justify="right")
    
    perf_table.add_row("BERT-base", "0.92", "0.94", "0.90")
    perf_table.add_row("RoBERTa-base", "0.93", "0.95", "0.91")
    perf_table.add_row("GPT-2", "0.89", "0.91", "0.87")
    
    # Таблица для ресурсов
    resource_table = Table(title="Ресурсы", show_header=True, header_style="bold blue")
    resource_table.add_column("Модель", style="cyan")
    resource_table.add_column("Память (GB)", style="magenta", justify="right")
    resource_table.add_column("Время (мин)", style="yellow", justify="right")
    resource_table.add_column("Параметры (M)", style="blue", justify="right")
    
    resource_table.add_row("BERT-base", "2.1", "45", "110")
    resource_table.add_row("RoBERTa-base", "2.3", "50", "125")
    resource_table.add_row("GPT-2", "1.8", "35", "117")
    
    # Создание панелей
    perf_panel = Panel(perf_table, title="[bold green]Метрики качества[/bold green]", border_style="green")
    resource_panel = Panel(resource_table, title="[bold blue]Требования к ресурсам[/bold blue]", border_style="blue")
    
    # Объединение в колонки
    console.print(Columns([perf_panel, resource_panel], equal=True, expand=True))

create_model_comparison()
```

## 6. Полезные функции для работы с таблицами


```python
# Универсальная функция для создания красивых таблиц
def create_rich_table(data, title="Таблица", column_styles=None, highlight_condition=None):
    """
    Создает красивую таблицу из данных
    
    Args:
        data: dict или list of dicts с данными
        title: заголовок таблицы
        column_styles: dict со стилями для колонок
        highlight_condition: функция для условного выделения строк
    """
    table = Table(title=title, show_header=True, header_style="bold magenta")
    
    if isinstance(data, dict):
        # Если данные в виде словаря
        for key, values in data.items():
            style = column_styles.get(key, "cyan") if column_styles else "cyan"
            table.add_column(key, style=style)
        
        # Определяем количество строк
        n_rows = len(next(iter(data.values())))
        for i in range(n_rows):
            row_data = [str(data[key][i]) for key in data.keys()]
            table.add_row(*row_data)
    
    elif isinstance(data, list) and len(data) > 0:
        # Если данные в виде списка словарей
        columns = list(data[0].keys())
        for col in columns:
            style = column_styles.get(col, "cyan") if column_styles else "cyan"
            table.add_column(col, style=style)
        
        for row in data:
            row_data = [str(row[col]) for col in columns]
            table.add_row(*row_data)
    
    return table

# Пример использования
experiment_data = {
    "Эксперимент": ["Baseline", "Data Aug", "Pretraining", "Fine-tuning"],
    "Accuracy": [0.85, 0.89, 0.94, 0.96],
    "Loss": [1.2, 0.9, 0.6, 0.4],
    "Время (мин)": [30, 45, 90, 120]
}

styles = {
    "Эксперимент": "cyan",
    "Accuracy": "green",
    "Loss": "red", 
    "Время (мин)": "yellow"
}

console.print(create_rich_table(experiment_data, "Результаты экспериментов", styles))
```

## 7. Экспорт в HTML и другие форматы


```python
# Экспорт таблицы в HTML
from rich.console import Console
from rich.table import Table
from io import StringIO

def export_table_to_html(table):
    """Экспорт Rich таблицы в HTML"""
    console = Console(file=StringIO(), force_terminal=True, width=100)
    console.print(table)
    return console.file.getvalue()

# Создание простой таблицы для экспорта
export_table = Table(title="Экспорт таблицы", show_header=True)
export_table.add_column("Колонка 1", style="cyan")
export_table.add_column("Колонка 2", style="green")
export_table.add_row("Значение 1", "Значение 2")

# Вывод HTML
html_output = export_table_to_html(export_table)
print("HTML вывод:")
print(html_output)
```

## Заключение

Библиотека `rich` предоставляет мощные инструменты для создания красивых таблиц в Python:

### Основные возможности:
- **Стилизация**: цвета, шрифты, выравнивание
- **Условное форматирование**: выделение строк на основе значений
- **Интеграция с Pandas**: легкая конвертация DataFrame
- **Панели и группировка**: организация нескольких таблиц
- **Экспорт**: возможность сохранения в различные форматы

### Ключевые классы:
- `Table` - основной класс для создания таблиц
- `Console` - для вывода и управления стилями
- `Panel` - для создания панелей
- `Columns` - для группировки элементов

### Полезные параметры:
- `style` - стиль текста и границ
- `justify` - выравнивание (left, center, right)
- `width` - ширина колонки
- `no_wrap` - запрет переноса строк
